```python
import pandas as pd
from etl import cargar_datos, resumen_paises

df_micro = cargar_datos("data/pisa2022.parquet")
df_macro = resumen_paises(df_micro)

print(f"Total de estudiantes analizados: {len(df_micro):,}")
print(f"Total de países/territorios: {df_macro['CNT'].nunique()}")
print("\n--- Promedios Globales ---")
print(f"Matemáticas: {df_macro['Matematicas_Avg'].mean():.2f}")
print(f"Lectura: {df_macro['Lectura_Avg'].mean():.2f}")
print(f"Ciencias: {df_macro['Ciencias_Avg'].mean():.2f}")
print(f"Ansiedad Matemática Promedio: {df_macro['Ansiedad_Avg'].mean():.2f}")
```

```python
for col in ['Matematicas_Avg', 'Lectura_Avg', 'Ciencias_Avg']:
    top = df_macro.sort_values(by=col, ascending=False).head(5)[['CNT', col]]
    print(f"\n=== Top 5 {col} ===")
    print(top.to_string(index=False))
```

```python
# Brecha Socioeconómica
escs_agg = df_micro.groupby('Nivel_Socioeconomico', observed=False)['Matemáticas'].mean()
print("=== Promedio por Cuartil Socioeconómico (Matemáticas) ===")
print(escs_agg)

# Brecha por Género
gender_agg = df_micro.groupby('Genero')[['Matemáticas', 'Lectura', 'Ciencias']].mean()
print("\n=== Promedio Global por Género ===")
print(gender_agg)
```

```python
top10_math = df_macro.sort_values(by='Matematicas_Avg', ascending=False).head(10)
print("=== Top 10 Matemáticas vs. Su Nivel de Ansiedad ===")
print(top10_math[['CNT', 'Matematicas_Avg', 'Ansiedad_Avg']].to_string(index=False))
```
